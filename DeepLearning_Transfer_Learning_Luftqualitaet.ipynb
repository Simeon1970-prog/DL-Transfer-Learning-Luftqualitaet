>### 1.3 Kurze wissenschaftliche Betrachtung

>**Relevanz:** Luftverschmutzung verursacht weltweit 4,2 Millionen vorzeitige Todesfälle pro Jahr [1]. In Europa überschreiten 97% der städtischen Bevölkerung die WHO-Grenzwerte für PM₂.₅ [2]. Zuverlässige Kurzzeit-Vorhersagen (24h) ermöglichen präventive Gesundheitswarnungen für vulnerable Bevölkerungsgruppen.

>**Ausgangspunkt – ML-Semesterarbeit:** Eine vorangegangene Arbeit mit klassischen ML-Methoden (XGBoost, Random Forest, Gradient Boosting, Ensemble) erreichte R² = 0.872. Die Analyse zeigte jedoch, dass 85% der Vorhersagekraft auf Autokorrelation durch Lag-Features basierte:

>$$Y(t+1) \approx \text{mean}(Y[t-24:t]) + \epsilon$$

>Ohne Lag-Features fiel R² auf 0.13 – das Modell lernte lediglich „morgen ≈ heute". Transfer zwischen Stationen war nicht möglich.

>**Problem:** Klassische ML-Modelle weisen zwei zentrale Limitationen auf: Erstens basiert ihre Vorhersagekraft überwiegend auf Autokorrelation – sie lernen keine echten Muster, sondern nur „morgen ≈ heute". Zweitens können sie nicht zwischen Stationen generalisieren, da jede Station eigene Schadstoffprofile aufweist [3]. Neue LANUK-Messstationen benötigen daher typischerweise 1–2 Jahre historische Daten, bevor zuverlässige ML-Prognosen möglich sind.

>**Forschungslücke:** Foundation Models bieten durch Pre-Training auf großen Datensätzen das Potential für Transfer Learning [4]. Ob dieser Ansatz auf Luftqualitätsdaten übertragbar ist, wurde bisher nicht systematisch untersucht.

>**Offene Fragen dieser Arbeit:**

>Kann ein Foundation Model für Zeitreihen auf Luftqualitätsdaten angewendet werden, um Transfer Learning zwischen Messstationen zu ermöglichen?

## 2. Überleitung von ML auf Deep Learning

>### 2.1 Das Autokorrelations-Problem

>Die ML-Semesterarbeit zeigte ein fundamentales Problem:

>| Konfiguration | R² |
|---------------|-----|
| Mit Lag-Features | 0.872 |
| Ohne Lag-Features | 0.13 |

>**85% der Vorhersagekraft basierte auf Autokorrelation.** Das Modell lernte lediglich: „morgen ≈ heute" [6].


---

>### 2.2 Warum Ausreißer wichtig sind

>| Ereignis | Beispiel | Problem für ML |
|----------|----------|----------------|
| Silvester-Feuerwerk | PM₁₀ > 200 µg/m³ | Autokorrelation bricht |
| Sahara-Staub | PM₁₀ > 150 µg/m³ | Kein lokales Muster |
| Inversionswetterlagen | NO₂-Spitzen | Meteorologie-abhängig |

>**Zentrale Erkenntnis:** Genau in den Momenten, wo Autokorrelation versagt (Extremereignisse), werden Vorhersagen am wichtigsten für Gesundheitswarnungen. Klassisches ML versagt hier.

---

>### 2.3 Anforderungen an eine bessere Lösung

>Aus den Erkenntnissen ergeben sich vier Strategien:

>1. **Kombinierter Index:** Ein kombinierter, gewichteter Index mit NO₂/O₃ Beimischung, der anschließend normiert wird
>2. **Längere Sequenzen:** Um Wochen- und Wettermuster zu erkennen
>3. **Multi-Station Training:** Erzwingt Generalisierung statt stationsspezifischer Autokorrelation
>4. **Anomalie-Behandlung:** Ein Foundation Model, das gut bis sehr gut auf Ausreißer reagieren kann

## 3. WCAQI – Weighted Composite Air Quality Index

>### 3.1 Herleitung

>Aus Kapitel 2: Um die Autokorrelation zu brechen, wird ein kombinierter, gewichteter Index mit NO₂/O₃ Beimischung benötigt, der anschließend normiert wird.

---

>### 3.2 WHO Air Quality Guidelines als Grundlage

>| Schadstoff | Grenzwert | Einheit | Gesundheitsrisiko |
|------------|-----------|---------|-------------------|
| PM2.5 | 5 | µg/m³ | Hoch |
| PM10 | 15 | µg/m³ | Hoch |
| NO2 | 10 | µg/m³ | Mittel-Hoch |
| O3 | 60 | µg/m³ | Mittel |
| SO2 | 40 | µg/m³ | Niedrig-Mittel |
| CO | 4000 | µg/m³ | Mittel |

>Quelle: WHO Air Quality Guidelines (2021) [1]

---

>### 3.3 Parameter

>**Gewichtungsfaktoren wᵢ (eigene Ableitung aus WHO-Gesundheitsrisiko) und Grenzwerte Gᵢ (EU 2024):**

>| Schadstoff | wᵢ | Gᵢ |
|------------|------|------------|
| PM2.5 | 0.30 | 25 µg/m³ |
| PM10 | 0.25 | 40 µg/m³ |
| NO2 | 0.20 | 40 µg/m³ |
| O3 | 0.15 | 120 µg/m³ |
| NO | 0.10 | 200 µg/m³|

>EU 2024 statt WHO 2021, da WHO-Grenzwerte zu 53-68% Capping führten.

---

>### 3.4 Formel

>**Normalisierung:**
>```
>Cᵢ_norm(t) = min(Cᵢ(t) / Gᵢ, 1.0) × 100
>```

>**WCAQI:**
>```
>WCAQI(t) = Σᵢ wᵢ × Cᵢ_norm(t)
>```

>wobei:
>- `Cᵢ(t)` = Konzentration des Schadstoffs i zum Zeitpunkt t
>- `Gᵢ` = Grenzwert für Schadstoff i (siehe 3.3)
>- `wᵢ` = Gewichtungsfaktor (siehe 3.3)
>- `Σᵢ wᵢ = 1.0`

---

>### 3.5 Zusätzlicher Vorteil: Differenzierbarkeit

>Während der Entwicklung wurde erkannt, dass WCAQI als gewichtete Summe vollständig differenzierbar ist – im Gegensatz zu LQI (max-Operator). Dies ist ein Bonus für Deep Learning. Diese Erkenntnis wird zur Kenntnis genommen, allerdings nicht weiter verfolgt, da der WCAQI ein "Mittel zum Zweck" ist und nicht Gegenstand dieser Arbeit.

## 4. Stand der Forschung: Time Series Foundation Models

## 5. MOMENT – Foundation Model Time Series

>### 5.1 Beschreibung
>MOMENT (MOdel for TimE SerieS) ist ein Foundation Model für Zeitreihenanalyse, entwickelt vom AutonLab der Carnegie Mellon University und vorgestellt auf der ICML 2024 [16].
>>#### 5.1.1 Die Architektur
>>Die Architektur basiert auf dem T5-Encoder [18]. Die Entwickler trainierten MOMENT auf dem Time-Series Pile (siehe 6.3). Dabei werden zufällige Teile der Zeitreihe verdeckt und das Modell lernt, diese aus dem Kontext zu ergänzen (Masked Time Series Modeling).
Der T5-Encoder ist im Grunde ein reiner Transformer-Encoder (wie die linke Hälfte des originalen Transformer von Vaswani et al. [17]).

>>#### 5.1.2 Der Encoder
>>Die Kernidee: T5 (Text-to-Text Transfer Transformer) hat eigentlich einen Encoder UND Decoder. MOMENT nutzt aber nur den Encoder-Teil, weil es keine Text-Generierung braucht, sondern Repräsentationen lernen soll.
>>
>>**Was der T5-Encoder macht:**
Die Eingabe (bei MOMENT: ein Zeitreihen-Segment von 512 Zeitpunkten) wird durch mehrere gestapelte Transformer-Blöcke geschickt. Jeder Block besteht aus Self-Attention (jeder Zeitpunkt "schaut" auf alle anderen Zeitpunkte und lernt, welche relevant sind) gefolgt von einem Feed-Forward-Netzwerk. Dazwischen liegen Layer "Normalization" und "Residual Connections".
>>
>>**Warum Encoder-only?**
>>Bei MOMENT geht es darum, eine gute Repräsentation der gesamten Eingabe-Zeitreihe zu erzeugen — nicht darum, Token für Token etwas zu generieren. Der Encoder sieht die komplette Eingabe bidirektional (vorwärts und rückwärts), was für das Masked Modeling ideal ist: Um einen verdeckten Abschnitt zu rekonstruieren, braucht das Modell Kontext von beiden Seiten.
Das unterscheidet sich von GPT-artigen Modellen (Decoder-only), die nur nach links schauen können. Für Zeitreihen-Rekonstruktion wäre das ein Nachteil.

>### 5.4 In- / Output

>>#### 5.4.1 Eingabeparameter

>>| Parameter | Wert |
|-----------|------|
| Sequenzlänge | 512 Zeitschritte (fest) |
| Channels | 1 (univariat) |
| Patch-Größe | 8 Zeitpunkte |
| Anzahl Patches | 64 |
| Normalisierung | z-Score erforderlich |

>>#### 5.4.2 Patching: Von Zeitpunkten zu Tokens

>>MOMENT nimmt nicht jeden einzelnen Zeitpunkt als eigenes Token — das wären 512 Token und damit zu viel für den Transformer. Stattdessen wird die Zeitreihe in **64 Patches à 8 Zeitpunkte** zerlegt.

>>**Warum Patching?** Self-Attention hat quadratische Komplexität in der Sequenzlänge. 512 Token würden 262.144 Vergleiche erfordern, 64 Patches nur 4.096 — das ist **64× effizienter**.

![Patching Pipeline](patching_pipeline.png)
>>*Abbildung 5.1: Patching — 512 Zeitpunkte werden in 64 Patches à 8 Werte zerlegt, linear eingebettet und als Token-Vektoren dem T5-Encoder [18] übergeben.*

>>Jeder Patch wird über eine lineare Projektion in den Embedding-Raum des T5-Encoders [18] projiziert. Ab diesem Punkt arbeitet der Transformer auf 64 Patch-Tokens — genau wie ein Sprachmodell auf Wort-Tokens arbeitet. Nur dass hier jedes "Wort" **8 Stunden** Luftqualitätsdaten repräsentiert.

>>#### 5.4.3 Univariat vs. Multivariat

>>MOMENT [16] ist ein **univariates** Modell — es verarbeitet immer nur **1 Variable** als Eingabe, eine eindimensionale Zeitreihe mit 512 Werten.

>>**Multivariat** wäre das Gegenteil: Alle Variablen gleichzeitig als Matrix einspeisen. Ein Zeitschritt hätte dann mehrere Werte und das Modell würde die Zusammenhänge zwischen den Variablen intern lernen (z.B. "wenn Wind zunimmt UND Temperatur sinkt, dann steigt PM10").
>>| | Multivariat (MOMENT kann das NICHT) | Univariat (so arbeitet MOMENT) |
|---|---|---|
| **Eingabe pro Zeitschritt** | `[PM10, PM2.5, NO2, O3, SO2, Temp, Wind, ...]` | `[WCAQI]` |
| **Kanäle** | 12 gleichzeitig | 1 |

>>### 5.4.4 Konsequenzen für das Training

>>**Pre-Training (Reconstruction):** Die 12 Komponenten (5 Luftqualität + 7 Meteorologie) pro Station werden als **12 separate Zeitreihen** behandelt. Jede wird einzeln in Segmente à 512 Zeitpunkte geschnitten und einzeln durch MOMENT geschickt. Das Modell sieht nie, dass PM10 und Temperatur zusammengehören — es lernt ausschließlich allgemeine Zeitreihen-Muster (Saisonalität, Trends, Autokorrelation).

![Univariate Verarbeitung](univariate_processing.png)
>>*Abbildung 5.2: Univariate Verarbeitung — Jede der 12 Komponenten wird einzeln durch MOMENT geschickt. Das Modell sieht keinen Zusammenhang zwischen den Variablen.*

>>Daher die 21.195 Segmente im Pre-Training: 18 Stationen × mehrere Komponenten × überlappende Fenster — aber alles Einzelsegmente.

>>**Fine-Tuning & Transfer (Forecasting):** Hier geht nur noch **1 Variable** rein und raus — der WCAQI:

>>| | Dimension | Beschreibung |
|---|---|---|
| **Input** | 512 × 1 | 512h WCAQI-Historie (z-Score normalisiert) |
| **Output** | 24 × 1 | 24h WCAQI-Forecast |

>>#### 5.4.5 Warum univariat?

>>Die Entwickler [16] haben MOMENT bewusst univariat designt — der Grund ist **Generalisierbarkeit**. Ein univariates Modell kann auf jede beliebige Zeitreihe angewendet werden, egal ob Luftqualität, Aktienkurse oder Herzfrequenz. Ein multivariates Modell müsste wissen, wie viele Variablen es gibt und was sie bedeuten.

>>| | Univariat (MOMENT) | Multivariat |
|---|---|---|
| **Eingabe** | 1 Variable × 512 Zeitpunkte | n Variablen × 512 Zeitpunkte |
| **Generalisierbarkeit** | Hoch — beliebige Zeitreihen | Niedrig — feste Variablenstruktur |
| **Inter-Variablen-Korrelation** | Nicht genutzt | Wird gelernt |
| **Beispiel** | WCAQI allein | PM10 + NO2 + Temp + Wind + ... |

>>**Nachteil:** Die Korrelationen zwischen Variablen (z.B. "Wind bläst Schadstoffe weg") werden nicht genutzt. Der WCAQI fängt das teilweise auf, weil er als gewichteter Index bereits mehrere Schadstoffe kombiniert — aber die meteorologischen Zusammenhänge gehen verloren.
>### 5.5 Modus-Varianten

>| Modus | Head | Zero-Shot* | Anwendung |
|-------|------|-----------|-----------|
| **Reconstruction** | PretrainHead (vortrainiert) | + | Imputation, Anomaly Detection |
| **Embedding** | Kein Head | + | Clustering, Feature-Extraktion |
| **Forecasting** | ForecastingHead (random) | - | Zeitreihen-Vorhersage |
| **Classification** | ClassificationHead (random) | - | Zeitreihen-Klassifikation |

>**Legende:** + = Zero-Shot möglich | - = Fine-Tuning erforderlich

>*Wichtig: Zwei Definitionen von "Zero-Shot"**

>Der Begriff "Zero-Shot" hat zwei unterschiedliche Bedeutungen:

>**Definition A – Native Zero-Shot (MOMENT Paper):**

>Das Modell direkt von HuggingFace laden und ohne jegliches Training verwenden. Dies funktioniert nur für Reconstruction und Embedding. Der ForecastingHead ist random initialisiert – ohne Training unbrauchbar.

>**Definition B – Transfer Learning Zero-Shot:**

>Das Modell wurde auf einer Quell-Domäne trainiert und wird auf einer neuen Ziel-Domäne ohne weitere Anpassung angewendet.

>| Begriff | Training auf Quelle? | Training auf Ziel? | Forecasting? |
|---------|---------------------|-------------------|--------------|
| Native Zero-Shot | Nein | Nein | Nein |
| Transfer Zero-Shot | Ja | Nein | Ja |

>### 6.3 Exkurs: Verlustfunktionen bei Regressionsaufgaben

Nach den schlechten MSE-Ergebnissen wurde recherchiert, welche Loss-Funktion besser für Luftqualitätsdaten geeignet ist [23].

>>#### 6.3.1 Überblick Verlustfunktionen für Regression

>>| Verlustfunktion | Empfindlichkeit gegenüber Ausreißern | Anwendung |
|-----------------|--------------------------------------|-----------|
| MSE (L2 Loss) | Hoch | Normalverteilte Daten ohne Ausreißer |
| MAE (L1 Loss) | Niedrig | Daten mit vielen Ausreißern |
| Huber Loss | Mittel (einstellbar) | Kompromiss zwischen MSE und MAE |

>>>##### 6.3.1.1 MSE (Mean Squared Error)

>>>$$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

>>>**Eigenschaften:**
>>>- Bestraft Ausreißer quadratisch — große Fehler dominieren den Gradienten
>>>- Gut bei normalverteilten Fehlern
>>>- Problematisch bei Luftqualitätsdaten (Silvester-Feuerwerk, Sahara-Staub)

>>>##### 6.3.1.2 MAE (Mean Absolute Error)

>>>$$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$$

>>>**Eigenschaften:**
>>>- Bestraft alle Fehler linear — Ausreißer haben keinen überproportionalen Einfluss
>>>- Robuster als MSE bei Daten mit extremen Werten
>>>- Gradient ist konstant (±1), unabhängig von der Fehlergröße
>>>- Nachteil: Nicht differenzierbar bei $y_i = \hat{y}_i$ (Gradient springt), was die Konvergenz nahe dem Optimum verlangsamen kann
>>>- Nachteil: Behandelt alle Fehler gleich — auch kleine Fehler werden nicht stärker optimiert

>>>##### 6.3.1.3 Huber Loss

>>>$$L_\delta(a) = \begin{cases} \frac{1}{2}a^2 & \text{für } |a| \leq \delta \\ \delta \cdot (|a| - \frac{1}{2}\delta) & \text{sonst} \end{cases}$$

>>>**Eigenschaften:**
>>>- Verhält sich quadratisch für kleine Fehler (wie MSE)
>>>- Verhält sich linear für große Fehler (wie MAE)
>>>- Parameter δ steuert den Übergang zwischen beiden Modi
>>>- Mit δ=2.0 werden Ausreißer weniger stark bestraft

>>#### 6.3.2 Entscheidung für Huber Loss

>>Luftqualitätsdaten enthalten regelmäßig Ausreißer durch:
>>- Silvester-Feuerwerk (PM10/PM2.5 Spitzen)
>>- Sahara-Staub Ereignisse
>>- Industrieunfälle
>>- Inversionswetterlagen

>>Diese Ereignisse sind **keine Messfehler**, sondern reale Phänomene. MSE würde das Modell zwingen, diese extremen Werte überzubetonen. Reines MAE würde zwar robust gegenüber Ausreißern sein, aber kleine Fehler im Normalbereich nicht präzise genug optimieren. Huber Loss mit δ=2.0 bietet den optimalen Kompromiss: kleine Fehler (|a| ≤ 2.0) werden quadratisch optimiert wie bei MSE, große Ausreißer (|a| > 2.0) werden linear behandelt wie bei MAE.

>>**Erkenntnis:** Wir sind auf dem richtigen Weg — wir probieren weiter mit Huber Loss.

>### 6.6 Schlussfolgerungen und Erkenntnisse

>>#### 6.6.1 Ursache-Wirkungs-Kette

>>1. **Ausgangspunkt:** Nur eine Station (EIFE) als Trainingsdaten
>>2. **Folge 1:** Zu wenig Trainingsdaten und geringe Variabilität
>>3. **Folge 2:** Extremereignisse (Silvester, Sahara-Staub) sind unterrepräsentiert
>>4. **Folge 3:** Modell lernt, dass Extremwerte selten sind
>>5. **Folge 4:** Systematische Unterschätzung (Bias –2.05)
>>6. **Ergebnis:** R² = 0.408 — nicht ausreichend für Transfer Learning

>### 8.2 Ausreißer-Behandlung (IQR-Capping)

![Outliers Before/After PM10](PreProcessing-ALLES/lanuv_batch_21/preprocessed/analysis/figures/outliers/outliers_before_after_PM10.png)

>**Was zeigt diese Grafik?**

>Das Beispiel PM10 zeigt den Effekt des IQR-Capping:
>- **Links (Before):** Originaldaten mit Ausreißern bis >150 µg/m³
>- **Rechts (After):** Nach Capping auf IQR-Grenze begrenzt
>- Die Zeitreihen-Kontinuität bleibt erhalten

>>#### 8.2.1 QR-Methode (Tukey 1977) [27]

>>$$IQR = Q3 - Q1$$
>>$$\text{Untere Grenze} = Q1 - 1{,}5 \times IQR$$
>>$$\text{Obere Grenze} = Q3 + 1{,}5 \times IQR$$

>>>##### 8.2.1.1 Warum 1,5 als Faktor?

>>>Der Faktor 1,5 wurde von Tukey empirisch gewählt und hat sich als guter Kompromiss zwischen Sensitivität und Robustheit erwiesen [28]:

>>>| Faktor | Grenze bei Normalverteilung | Anteil Ausreißer |
|--------|----------------------------|------------------|
| 1,0 | ±2,025σ | ~4,3% |
| **1,5** | **±2,7σ** | **~0,7%** |
| 3,0 | ±4,05σ | ~0,0002% |

>>>##### 8.2.1.2 Problem bei niedrigem IQR

>>>Bei Stationen mit sehr stabilen Werten (z.B. EIFE NO2) kann der IQR nahe 0 sein. In diesem Fall wird ein Fallback auf das 99,5-Perzentil verwendet:
>>>
>>>```python
>>>if IQR < 0.5:
>>>    cap_value = df[component].quantile(0.995)  # Fallback
>>>else:
>>>    cap_value = Q3 + 1.5 * IQR  # Standard
>>>```

>>#### 8.2.2 Warum Capping statt Trimming?

>>**Trimming** bedeutet, dass Ausreißer komplett aus dem Datensatz entfernt werden. Der Messwert wird gelöscht und hinterlässt eine Lücke in der Zeitreihe. Dies widerspricht zudem der Tatsache, dass gerade beim Feinstaub (PM) die Ausreißer (Silvester und Sahara Staub...) besonders wichtig sind.

>>**Capping** bedeutet, dass Ausreißer auf einen Grenzwert begrenzt werden. Der Messwert bleibt erhalten, wird aber auf die obere oder untere Grenze gesetzt.


>>| Aspekt | Trimming (Löschen) | Capping (Begrenzen) |
|--------|-------------------|---------------------|
| Was passiert? | Wert wird gelöscht → NaN | Wert wird auf Grenze gesetzt |
| Datenverlust | Ja | Nein |
| Zeitreihen-Kontinuität | Unterbrochen (Lücken) | Erhalten (keine Lücken) |

>>#### 8.2.3 Warum ist das für MOMENT relevant?

>>MOMENT benötigt in allen Trainingsphasen **lückenfreie Daten** [31]:

>>| Phase | Modus | Lücken erlaubt? | Begründung |
|-------|-------|-----------------|------------|
| **Pre-Training** | Reconstruction | Nein | Vollständige Daten als Ground Truth für Loss-Berechnung |
| **Fine-Tuning** | Forecasting | Nein | ForecastingHead erwartet vollständige Inputs |
| **Imputation (Inference)** | Reconstruction | Ja | Zero-Shot, kein Training, keine Ground Truth nötig |

>>Beim Pre-Training maskiert MOMENT **selbst** zufällige Patches der vollständigen Zeitreihe und lernt diese zu rekonstruieren. Der Loss wird zwischen Rekonstruktion und Original berechnet - dafür muss das Original vollständig sein.

>>**Entscheidung für Capping:** Trimming würde neue Lücken erzeugen, die dann wieder imputiert werden müssten - ein Widerspruch zur gerade durchgeführten Imputation. Capping erhält die Kontinuität und bewahrt gleichzeitig die Information, dass zu diesem Zeitpunkt ein Extremereignis stattfand (z.B. Silvester-Feuerwerk, Sahara-Staub).

>### 8.3 Untersuchung von Verzerrungen durch Imputationen - MCAR vs. MAR Diagnostik

>>### 8.3.1 Wissenschaftliche Bedeutung

>>Die Klassifikation des Missing Data Mechanism ist fundamental für die Validität jeder Imputation. Rubin (1976) [30] führte diese Taxonomie ein:

>>| Mechanismus | Warum fehlen Daten? | Beispiel | Konsequenz |
|-------------|---------------------|----------|------------|
| **MCAR** | Rein zufällig | Stromausfall | ✅ Imputation sicher |
| **MAR** | Abhängig von **gemessenen** Variablen | Wochenend-Ausfälle | ✅ Imputation sicher |
| **MNAR** | Abhängig vom **fehlenden Wert selbst** | Sensor fällt bei hoher Belastung aus | ❌ Imputation verzerrt! |

>>**Das Problem bei MNAR:** Wenn Sensoren gerade dann ausfallen, wenn die Schadstoffwerte hoch sind, würde Imputation die hohen Werte systematisch **unterschätzen**. Das Modell lernt dann falsche Muster.

![MCAR/MAR Diagnostik](PreProcessing-ALLES/lanuv_batch_21/preprocessed/analysis/gap_analysis/fig_mcar_mar_diagnostik.png)

>>### 8.3.2 Testmethodik: Approximation von Little's MCAR Test [29]

>>Für jede Komponente wird ein **Missing-Indikator** erstellt:

>>$$M_i = \begin{cases} 1 & \text{wenn } X_i \text{ fehlt} \\ 0 & \text{wenn } X_i \text{ vorhanden} \end{cases}$$

>>Die **Point-Biserial Korrelation** zwischen Missing-Indikator und anderen Variablen:

>>$$r_{pb} = \frac{\bar{X}_1 - \bar{X}_0}{s_X} \cdot \sqrt{\frac{n_1 \cdot n_0}{n^2}}$$

>>### 8.3.3 Interpretation
>>- $p < 0.05$: Signifikant → MCAR abgelehnt → MAR
>>- $p \geq 0.05$: Nicht signifikant → MCAR

>>**MCAR-Test Ergebnisse (Zelle 20):**

>>| Metrik | Wert |
|--------|------|
| Anzahl Tests | 246 |
| Signifikante Korrelationen | 110 (44,7%) |

>>**Ergebnis:** MCAR abgelehnt → **Klassifikation: MAR**

>>Die hohe Anzahl signifikanter Korrelationen (44,7%) zeigt, dass MCAR **nicht** gilt. Die fehlenden Werte hängen von anderen beobachteten Variablen ab (z.B. Wartungszyklen, Wochenenden) - typisches MAR-Muster.

>>### 8.3.3 Schlussfolgerung: 
>>**Schlussfolgerung:** Die Daten sind MAR, nicht MNAR. Die verwendeten Imputationsmethoden führen zu **keiner systematischen Verzerrung**.

>## 9.1 Die drei Phasen des Trainings

>Foundation Models wie MOMENT [16] werden nicht direkt auf die Zielaufgabe trainiert. Stattdessen durchlaufen sie ein mehrstufiges Training, das allgemeines Wissen schrittweise auf die spezifische Aufgabe fokussiert — ein Prinzip, das Pan & Yang als "Transfer Learning" formalisiert haben [4].

>| Phase | Task | Ziel | Analogie (GPT) |
|-------|------|------|----------------|
| **Pre-Training** | Reconstruction (Self-Supervised) | Allgemeine Zeitreihen-Repräsentationen lernen | Texte aus dem Internet lesen |
| **Fine-Tuning** | Forecasting (Supervised) | WCAQI-Vorhersage auf bekannten Stationen | Auf Frage-Antwort-Paaren trainieren |
| **Transfer Learning** | Forecasting (wenig/keine Daten) | Vorhersage auf **ungesehenen** Stationen | Ein neues Fachgebiet verstehen |

>>### 9.1.1 Phase 1: Pre-Training (Reconstruction Task / Modus)

>>Das Backbone lernt hier "die Sprache der Luftqualitäts-Zeitreihen" zu verstehen. Durch das Rekonstruieren maskierter Patches lernt es grundlegende Muster wie Tageszyklen von NO2 (Berufsverkehr), saisonale Schwankungen von O3 (Sommer höher), Korrelationen zwischen Komponenten, und meteorologische Einflüsse. Das ist wie wenn ein Kind erst lernt, Buchstaben und Wörter zu erkennen, bevor es Sätze bilden kann.

>>### 9.1.2 Phase 2: Fine-Tuning (Forecasting Task / Modus)

>>Hier lernt das Modell, aus den erkannten Mustern Vorhersagen abzuleiten. Der ForecastingHead lernt: "Wenn ich dieses Muster sehe, dann kommt wahrscheinlich das." Das ist wie wenn das Kind, das jetzt lesen kann, lernt, den nächsten Satz einer Geschichte vorherzusagen.

>>### 9.1.3 Phase 3: Transfer Learning (Forecasting Task / Modus)
>> In dieser Phase soll das System auf ungesehen Daten, anahnd der erlernten "Sprache der Luftqualitätsdaten" und der erkannten Wahrscheinlichkeiten eine Vorhersage treffen. Das eigentliche Transfer Learning.

>>### 9.1.4 Warum DIESE Phasen notwendig sind

>>Ohne Pre-Training auf LANUK-Daten hätte das Backbone nur allgemeine Zeitreihen-Muster gekannt (aus dem ursprünglichen MOMENT Training auf Energie, Finanzen, etc.) [16], aber keine Luftqualitäts-spezifischen Muster. Ohne Fine-Tuning wäre der ForecastingHead random und könnte die gelernten Repräsentationen nicht in Vorhersagen umwandeln.

>>**Analogie:** Wie ein Kind erst lesen lernt (Pre-Training), dann Textverständnis entwickelt (Fine-Tuning) und schließlich ein neues Buch versteht, ohne es vorher gelesen zu haben (Transfer Learning).

>>Dieses dreistufige Paradigma folgt dem etablierten Transfer-Learning-Ansatz [4]: Erst breites Wissen aufbauen, dann domänenspezifisch anpassen, schließlich auf neue Aufgaben transferieren. Die zugrunde liegende Transformer-Architektur basiert auf dem Attention-Mechanismus von Vaswani et al. [17], der auch bei MOMENT [16] zum Einsatz kommt. Die Idee des einheitlichen Pre-Training → Fine-Tuning Paradigmas wurde durch Raffel et al. mit dem T5-Modell systematisch untersucht [18].

>>### 9.1.5 Warum nicht direkt trainieren?

>>Klassische ML-Modelle (Random Forest, XGBoost) werden direkt auf den Zieldaten trainiert und benötigen typischerweise 1–2 Jahre historische Daten pro Station [3]. Der Foundation-Model-Ansatz ermöglicht:

>>1. **Wissenstransfer** zwischen Stationen — gelerntes Muster von 18 Stationen hilft bei ungesehenen Stationen [4]
>>2. **Dateneffizienz** — schon mit 10% der Target-Daten brauchbare Vorhersagen
>>3. **Schnelle Deployment-Zeiten** — neue Station in Wochen statt Jahren einsatzbereit

>>**Differential Learning Rate:** Der Backbone wird mit 100× niedrigerer LR als der Head trainiert. So bleibt das Pre-Training-Wissen erhalten, während der Head schnell die neue Aufgabe lernt. Dieses Prinzip des schichtweisen Lernraten-Managements ist ein etabliertes Verfahren im Transfer Learning [4].

>>Die zwei Kurven zeigen den 100×-Unterschied zwischen Head-LR (1e-3, oben) und Backbone-LR (1e-5, unten). >>**Huber Loss statt MSE:** Huber Loss kombiniert die Vorteile von MAE (robust gegen Ausreißer) und MSE (glatter Gradient nahe Null). Mit δ=2.0 werden extreme Fehlvorhersagen weniger bestraft als bei reiner MSE [23].

# 10 Diskussion

>## 10.1 Interpretation der Ergebnisse

>Die experimentellen Ergebnisse demonstrieren, dass MOMENT (Goswami et al., 2024 [16]) erfolgreich auf die Domäne der Luftqualitätsvorhersage transferiert werden kann. Das dreistufige Training (Pre-Training → Fine-Tuning → Transfer Learning) führte zu folgenden zentralen Erkenntnissen.
----
>Damit sind die in Kapitel 1 formulierten Leitfragen positiv beantwortet: Das Modell liefert ab Tag 1 nutzbare Prognosen für unbekannte Stationen (R² = 0,676), ohne auf deren historische Daten trainiert zu werden. Die Kernhypothese — eine Reduktion der Deployment-Zeit von 1–2 Jahren auf 2–4 Wochen — wird nicht nur bestätigt, sondern übertroffen: Zero-Shot-Transfer erfordert keine Anlaufphase. Mit kontrolliertem Fine-Tuning (EXP A, LR = 1e-6) verbessert sich die Performance monoton bis R² = 0,863. Die folgenden Abschnitte differenzieren dieses Ergebnis nach Stationstyp, Methodik und Limitation.
----


>>### 10.1.1 Zero-Shot-Fähigkeit als Basislinie

>>Die Zero-Shot-Vorhersage erreichte einen durchschnittlichen R²-Wert von **0,676** über alle drei Zielstationen — ohne jegliche stationsspezifische Trainingsdaten. Die Varianz zwischen den Stationen (R² 0,625–0,702) spiegelt die unterschiedliche Übertragbarkeit wider: Industrielle Standorte profitieren stärker vom Transfer als suburbane Stationen, da die 13 Trainingsstationen überwiegend urban geprägt sind.

>>### 10.1.2 Das Catastrophic-Forgetting-Problem und seine Lösung

>>Bei der Standard-Learning-Rate (5e-5) verschlechterten sich die Few-Shot-Szenarien bei **allen drei Stationen** gegenüber der Zero-Shot-Basislinie — ein paradoxes Ergebnis, bei dem mehr Daten zu schlechterer Performance führten (Catastrophic Forgetting; McCloskey & Cohen, 1989 [31]).

>>Experiment A bestätigte, dass eine 50-fach niedrigere Learning Rate (1e-6) dieses Problem **vollständig eliminiert**:

>>| Metrik | Basis (LR=5e-5) | EXP A (LR=1e-6) | Δ |
|--------|:---------------:|:----------------:|:---:|
| Avg R² (alle Szenarien) | 0,655 | 0,707 | **+0,052** |
| Catastrophic Forgetting | 3/3 Stationen | 0/3 Stationen | **eliminiert** |
| Bester Einzelwert | 0,778 (DMD2) | 0,863 (DMD2) | **+0,085** |
| Monotone Verbesserung | ✗ | ✓ (ZS→FS10→FS25→FT steigend) | **hergestellt** |

>>Experiment B (Backbone-Freeze, LR=5e-5) zeigte dagegen nur minimale Effekte (Δ R² ≈ 0,000): Der Forecasting Head allein hatte nicht genügend Kapazität für stationsspezifische Muster.

>>### 10.1.3 Stationstyp-abhängige Performance

>>| Station | Typ | Zero-Shot R² | Bester R² (EXP A) | Δ | Optimale Strategie |
|---------|-----|:----------:|:-----------------:|:---:|-------------------|
| DMD2 (Dortmund-Eving) | Industrial | 0,702 | **0,863** | +0,161 | Full Fine-Tuning |
| LOER (Düsseldorf-Lörick) | Urban | 0,702 | **0,737** | +0,035 | Few-Shot 25% |
| SHW2 (Schwerte) | Suburban | 0,625 | **0,657** | +0,032 | Few-Shot 25% |

>>Die industrielle Station profitiert massiv vom Fine-Tuning (+0,161), da regelmäßigere Emissionsmuster besser lernbar sind. Bei urbanen/suburbanen Stationen zeigt bereits der Zero-Shot-Transfer die beste Kosten-Nutzen-Relation.

>## 10.2 Was MOMENT kann

>| Fähigkeit | Evidenz | Metriken |
|-----------|---------|----------|
| **Domänen-Transfer** ohne AQ-Daten im Pre-Training | *Time Series Pile* enthält keine Luftqualitätsdaten; dennoch erfolgreiche Domain-Adaptation | R² ≈ 0,67 Zero-Shot |
| **Sofortige Einsatzfähigkeit** neuer Stationen | Tag 1 ohne lokale Daten nutzbar | Skill vs. Persistence: 0,52–0,60; Dir. Acc.: 64–67% |
| **Kontrolliertes Transfer Learning** | Monoton steigende Performance mit EXP A | ZS → FS10 → FS25 → FT durchgängig besser |
| **Schwellenwertbasierte Warnung** | Ausreichend für binäre Entscheidungen (WCAQI > 50? > 100?) | Dir. Acc. 64–76%; Skill 0,55–0,79 (EXP A) |

>## 10.3 Was MOMENT nicht kann

>>### 10.3.1 Architektonische Limitationen

>>| Limitation | Bedeutung | Neuere Modelle als Kontrast |
|------------|-----------|----------------------------|
| **Keine Kovariablen** | Temperatur, Wind, CAMS-Vorhersagen werden ignoriert; Inversionen, Föhn, Starkwind bleiben unsichtbar | Chronos-2 [32]: native Kovariablen via Group Attention (past-only + known future + kategorisch) |
| **Keine Konfidenzintervalle** | Nur Punktprognosen; keine Aussage „WCAQI > 100 mit 90% Wahrscheinlichkeit" möglich | Chronos-2: 21 Quantile (0,01–0,99); COSMIC: 9 Quantile (0,1–0,9) |
| **Kein multivariate Forecasting** | PM₁₀, NO₂, O₃, PM₂.₅ nicht simultan vorhersagbar; Wechselwirkungen (NOₓ-O₃-Gleichgewicht) nicht modelliert | Chronos-2: beliebig viele Zielgrößen gleichzeitig |
| **Keine räumliche Generalisierung** | Jede Station unabhängig; kein Windtransport, keine Interpolation zwischen Stationen | GNN-Ansätze (MSTGAN, GLA-Net) modellieren räumliche Abhängigkeiten |

>>### 10.3.2 Evaluations-Limitationen

>>| Limitation | Beschreibung |
|------------|-------------|
| **Begrenzte Stationsauswahl** | 3 von 61 LANUK-Stationen; keine verkehrsnahen oder ländlichen Hintergrundstationen evaluiert |
| **Keine operative Validierung** | Rein retrospektiv; Sensorausfälle, Datenlücken, Latenzzeiten nicht berücksichtigt |
| **Keine saisonale Differenzierung** | Performance bei Extremereignissen (Saharastaub, Inversionen, Sommersmog) ungeprüft |

>>### 10.3.3 Regulatorische Konsequenz

>>Die EU-AAQD 2024/2881 fordert kurzfristige Aktionspläne bei prognostizierten Grenzwertüberschreitungen (Art. 20):

>>| Anforderung (AAQD) | Status | Bewertung |
|---------------------|:------:|-----------|
| 24h-Vorhersage von Luftqualitätsindizes | ✅ | R² 0,62–0,86 |
| Quantifizierte Überschreitungswahrscheinlichkeit | ❌ | Keine probabilistischen Outputs |
| Integration externer Vorhersagemodelle (CAMS) | ❌ | Keine Kovariablen-Unterstützung |

>>→ Das System kann als **informative Entscheidungshilfe**, jedoch nicht als regulatorisch belastbare Vorhersage im Sinne der AAQD eingesetzt werden.

>## 10.4 Was noch getan werden muss

>| Prio | Maßnahme | Methode | Erwarteter Nutzen |
|:----:|----------|---------|-------------------|
| **1** | **Unsicherheitsquantifizierung** | Conformal Prediction (Achour et al., 2025 [33]) + ACI (Gibbs & Candès, 2021 [34]) | Warnungen: „WCAQI > 100 mit 90% Wahrscheinlichkeit"; AAQD-Compliance |
| **2** | **Kovariablen-Integration** | **Weg A:** Modellwechsel zu Chronos-2 (nativ multivariat + probabilistisch); **Weg B:** MOMENT als Bias-Korrektur für CAMS | Meteorologische Informationen nutzen; Inversionswetterlagen erkennen |
| **3** | **Prospektive Validierung** | 12 Monate Echtbetrieb parallel zum Monitoring an Pilotstation (SHW2) | Nachweis der Übertragbarkeit retrospektiver Metriken auf Echtzeit |
| **4** | **Erweiterte Evaluation** | 10–15 Stationen (inkl. Traffic, Rural); saisonale Aufschlüsselung | Generalisierbarkeit belegen; Schwächen bei Extremereignissen identifizieren |

>**Zu Priorität 1:** Conformal Prediction ist modellunabhängig und erfordert keine Änderung des MOMENT-Modells. Die Zero-Shot-Fähigkeit bietet einen Vorteil: Da keine Trainingsdaten benötigt werden, stehen mehr Daten für die Kalibrierung der konformen Intervalle zur Verfügung (Achour et al., 2025 [33]).

>**Zu Priorität 2:** Chronos-2 (Ansari et al., 2025 [32]) löst architektonisch sowohl das Kovariablen- als auch das Konfidenzintervall-Problem (21 Quantile nativ). Es unterstützt past-only Kovariablen (historische Windmessungen), known future Kovariablen (DWD-Wettervorhersagen) und kategorische Variablen. Alternativ kann MOMENT als Post-Processing-Schicht auf CAMS-Vorhersagen operieren (Bias-Korrektur), ohne Architekturänderung.

# 11 Zusammenfassung und Ausblick

>## 11.1 Wissenschaftlicher Beitrag

>Diese Arbeit hat gezeigt, dass Foundation Models für Zeitreihen — konkret das MOMENT-Modell (Goswami et al., 2024 [16]) mit 385 Millionen Parametern — erfolgreich auf die Domäne der Luftqualitätsvorhersage transferiert werden können. Die wesentlichen Beiträge sind:
---

>Die zentrale Forschungsfrage — ob Transfer Learning Luftqualitätsprognosen für neue Stationen ohne historische Daten ermöglicht — ist eindeutig positiv beantwortet. Die Kernhypothese einer Reduktion der Deployment-Zeit von 1–2 Jahren auf 2–4 Wochen wurde übertroffen: Zero-Shot-Prognosen sind ab dem ersten Betriebstag verfügbar (R² = 0,676), mit monotoner Verbesserung bei zunehmender Datenverfügbarkeit (bis R² = 0,863 bei Full Fine-Tuning). Im Einzelnen:

---

>>### 11.1.1 Dreistufiges Trainingsparadigma für WCAQI-Vorhersage
>> Das Pre-Training → Fine-Tuning → Transfer Learning-Pipeline wurde erstmals auf den Weighted Combined Air Quality Index (WCAQI) angewendet. Das Fine-Tuned-Modell erreichte im Zero-Shot-Transfer auf drei unbekannte Stationen einen durchschnittlichen R²-Wert von 0,676 — ohne jegliche stationsspezifische Daten.

>>### 11.1.2 Identifikation und Lösung von Catastrophic Forgetting beim Transfer Learning
>> Die zentrale Entdeckung dieser Arbeit: Bei Standard-Learning-Rate (5e-5) verschlechterte sich die Vorhersage mit zunehmenden Target-Daten bei allen drei Stationen gegenüber dem Zero-Shot — ein paradoxes Ergebnis, das als Catastrophic Forgetting (McCloskey & Cohen, 1989 [31]) bekannt ist. Experiment A (LR = 1e-6, 50-fach reduziert) eliminierte dieses Problem vollständig und stellte monotone Verbesserung her: Jedes zusätzliche Datenquantum führte zu besserer Performance (Zero-Shot R² 0,676 → Few-Shot 10% R² 0,703 → Few-Shot 25% R² 0,719 → Full FT R² 0,729). Der beste Einzelwert lag bei R² = 0,863 (DMD2, EXP A Full FT).

>>### 11.1.3 Experimenteller Nachweis der Überlegenheit kontrollierter Adaptation gegenüber Backbone-Freeze
>> Experiment B (vollständiges Einfrieren des Backbone) zeigte keine signifikante Verbesserung gegenüber der Basiskonfiguration (Δ R² ≈ 0,000). Dies belegt, dass eine kontrollierte Anpassung des gesamten Modells mit niedriger Learning Rate der strikten Trennung zwischen fixem Backbone und trainierbarem Head vorzuziehen ist.

>>### 11.1.4 Stationstyp-abhängige Transferierbarkeit
>> Die Ergebnisse zeigen eine klare Abhängigkeit vom Stationstyp: Industrielle Stationen (DMD2, R² bis 0,863) profitieren stärker vom Transfer als suburbane Standorte (SHW2, R² bis 0,657). Dies deutet auf eine Korrelation zwischen den im Pre-Training gelernten Mustern und urban-industriellen Emissionsprofilen hin.

---

>## 11.2 Weiteres Vorgehen

>Die in Kapitel 10.3 identifizierten Limitationen definieren die nächsten Schritte, priorisiert nach Umsetzbarkeit:

>| Prio | Maßnahme | Adressierte Limitation | Zeithorizont |
|:----:|----------|----------------------|:------------:|
| **1** | Conformal Prediction | Keine Konfidenzintervalle | Sofort umsetzbar |
| **2** | CAMS-Bias-Korrektur | Keine Kovariablen | 3–6 Monate |
| **3** | Chronos-2 Evaluation | Univariater Ansatz | 6–12 Monate |
| **4** | LoRA-Adapter | Skalierbarkeit (385M Parameter pro Station) | 3–6 Monate |
| **5** | UBA-Generalisierung | Nur 3 von 61 Stationen evaluiert | 6–12 Monate |

>**Priorität 1** ist sofort umsetzbar, weil Conformal Prediction modellunabhängig als Post-Processing auf die bestehenden Punktprognosen aufgesetzt wird (Achour et al., 2025 [33]). 
**Priorität 2** nutzt das bestehende MOMENT-Modell als lokale Feinkorrektur für CAMS-Regionalvorhersagen (~10–20 km Auflösung [2]) — die Modellarchitektur bleibt unverändert. 
**Priorität 3** evaluiert Chronos-2 (Ansari et al., 2025 [32]) als alternatives Foundation Model mit nativer Kovariablen-Unterstützung und probabilistischer Ausgabe (21 Quantile). 
**Priorität 4** reduziert den Speicherbedarf für die Skalierung auf 61 LANUV-Stationen von ~82 GB auf ~4,4 GB durch parameter-effizientes Fine-Tuning (Hu et al., 2022 [35]; Gupta et al., 2024 [36]).

>Priorität 1 und 2 lassen sich mit dem bestehenden Modellartefakt (`finetuning_best.pt`, 1,34 GB) und der validierten EXP-A-Konfiguration direkt umsetzen. Die Prioritäten 3–5 bilden den Kern der geplanten Masterarbeit.

>## 11.3 Projektbezogene Lessons Learned

>>### 11.3.1 Learning Rate ist der kritischste Hyperparameter beim Transfer Learning mit Foundation Models
>> Die Differenz zwischen funktionierendem und nicht funktionierendem Transfer lag nicht in der Architektur, nicht in der Datenmenge und nicht in der Trainingsdauer — sondern in einem einzigen Hyperparameter. Die Standard-Learning-Rate (5e-5) aus der MOMENT-Dokumentation zerstörte das vortrainierte Wissen; erst die 50-fache Reduktion auf 1e-6 stellte stabilen Transfer her.

>>### 11.3.2 Zero-Shot ist eine ernst zu nehmende Basislinie
>> Bei 2 von 3 Stationen war die Zero-Shot-Vorhersage besser als jede Form von Transfer Learning mit Standard-LR — einschließlich Full Fine-Tuning mit 70% der verfügbaren Daten. Diese Beobachtung hat direkte praktische Relevanz: Im kommunalen Einsatz sollte eine neue Station zunächst im Zero-Shot-Modus betrieben werden, bevor Transfer Learning mit validierten Hyperparametern erfolgt.

>>### 11.3.3 Backbone-Freeze allein reicht nicht
>> Die naheliegende Lösung — den vortrainierten Backbone einzufrieren und nur den Forecasting Head zu trainieren (EXP B) — brachte keine Verbesserung. Der Head allein hatte nicht genügend Kapazität, um stationsspezifische Muster zu lernen. Die Kombination aus trainierbarem Gesamtmodell und sehr niedriger Learning Rate war die effektivere Strategie.

>>### 11.3.4 Stationstypen sind nicht austauschbar
>> Die R²-Differenz zwischen DMD2 (Industrial, 0,863) und SHW2 (Suburban, 0,657) bei identischer Konfiguration zeigt, dass Transferierbarkeit vom Stationscharakter abhängt. Ein auf urban-industriellen Stationen trainiertes Modell generalisiert besser auf ähnliche Standorte. Für kleinstädtische und ländliche Stationen könnte ein heterogenerer Trainingsdatensatz oder stationstyp-spezifisches Fine-Tuning nötig sein.

>>### 11.3.5 Reproduzierbarkeit durch vollständige Dokumentation
>> Die systematische Protokollierung aller Trainingsläufe (RUN_ID: `20260221_173521_147aa291 `, VM-Spezifikation, Hyperparameter, Timing, Artefakte) ermöglicht die exakte Reproduktion aller Ergebnisse. Diese Praxis hat sich als unverzichtbar erwiesen, insbesondere bei der nachträglichen Analyse des Catastrophic-Forgetting-Phänomens.
